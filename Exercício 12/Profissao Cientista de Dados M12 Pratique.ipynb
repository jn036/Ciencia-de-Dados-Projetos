Durante a exploração de vocês podem utilizar mais de 3 gráficos mas deixem na entrega final apenas os 3 mais assertivos que contenham a informação da mensagem que querem passar, fiquem livres para utilizar as bibliotecas que quiserem para os gráficos.

Justifiquem a escolha dos visuais e despertem o lado analítico de vocês com uma análise acerca do descoberto através dos gráficos. Lembrem o que aprendemos na aula, devemos ser assertivos na nossa mensagem.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. CRIANDO O BANCO DE DADOS
data = {
    'Nome': ['Alice', 'Joao', 'Charlie', 'David', 'Eva', 'Diego', 'Denize', 'Claudio'],
    'Idade': [25, 30, 35, 40, 45, 60, 22, 24],
    'Profissão': ['Engenheiro', 'Médico', 'Professor', 'Advogado', 'Médico', 'Engenheiro', 'Estudante', 'Estudante'],
    'Salário': ['4500', '8000', '5000', '10000', '12000', '15000', '12000', '1500'],
    'Limite_Credito': ['2500', '4000', '4000', '1000', '10000', '2000', '500', '250'],
    'Historico_Inadimplencia': ['0', '0', '0', '1', '0', '1', '0', '1'],
    'Estado_Civil': ['Casamento', 'Casamento', 'Solteiro', 'Solteiro', 'Casamento', 'Solteiro', 'Solteiro', 'Solteiro'],
    'Imovel_Proprio': ['0', '0', '0', '1', '1', '1', '0', '0']
}

df = pd.DataFrame(data)

# 2. CONVERSÃO DE DADOS
# Convertendo colunas numéricas que estavam como texto para números
df['Salário'] = pd.to_numeric(df['Salário'])
df['Limite_Credito'] = pd.to_numeric(df['Limite_Credito'])
df['Idade'] = pd.to_numeric(df['Idade'])

# Configurando o visual dos gráficos
sns.set_theme(style="whitegrid")

# --- GRÁFICO 1: SALÁRIO VS LIMITE ---
# Usando um 'scatterplot' (gráfico de pontos) para ver se quando um sobe, o outro também sobe.
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Salário', y='Limite_Credito', hue='Profissão', s=100)
plt.title('Gráfico 1: Relação entre Salário e Limite de Crédito')
plt.show()

# --- GRÁFICO 2: HISTÓRICO DE INADIMPLÊNCIA ---
# Usando um 'barplot' (barras) para comparar a média de limite de quem deve vs quem não deve.
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Historico_Inadimplencia', y='Limite_Credito', palette='viridis')
plt.title('Gráfico 2: Impacto da Inadimplência no Limite (0=Limpo, 1=Sujo)')
plt.show()

# --- GRÁFICO 3: IDADE VS LIMITE ---
# Usando um 'regplot' que desenha uma linha de tendência para ver se a idade ajuda no limite.
plt.figure(figsize=(10, 6))
sns.regplot(data=df, x='Idade', y='Limite_Credito', color='blue')
plt.title('Gráfico 3: A Idade influencia no Limite de Crédito?')
plt.show()
```
