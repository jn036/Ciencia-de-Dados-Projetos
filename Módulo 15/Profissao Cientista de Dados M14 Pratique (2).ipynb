# 1 - Carregue a base, verifique se os tipos de dados estão de acordo com cada coluna e caso não estejam realize a transformação dos tipos de dados.

```python
# Definindo as colunas que devem ser numéricas
colunas_numericas = ['Pagamento_Mensal', 'Total_Pago', 'Tempo_como_Cliente', 'Idoso']

# Convertendo para numérico (errors='coerce' transforma valores inválidos em NaN para tratamento posterior)
for col in colunas_numericas:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Verificando se a transformação foi aplicada
print(df.dtypes)
```

# 2 - Esse exercício faremos por etapas:
A) Verifique se temos colunas com dados faltantes e traga a % dos dados faltantes por coluna.

```python
# Calculando o percentual de valores ausentes por coluna
missing_percent = (df.isnull().sum() / len(df)) * 100

# Filtrando apenas as colunas que possuem nulos e ordenando do maior para o menor
missing_percent = missing_percent[missing_percent > 0].sort_values(ascending=False)

print("Percentual de dados faltantes por coluna:")
print(missing_percent)
```

B) Exclua nessa etapa as linhas das colunas com dados faltantes que você analisar que esse método é o ideal. Justifique o motivo de estar excluindo essas linhas.

```python
#A coluna Churn é nossa variável alvo. Caso ela esteja faltando, a linha não pode ser usada para modelagem. Portanto, devemos excluir todas as linhas onde Churn é NaN, pois não sabemos se o cliente cancelou ou não.

#As colunas Pagamento_Mensal e Total_Pago possuem muitos valores ausentes (mais de 30%). A exclusão de todas essas linhas descartaria grande parte da base, o que seria prejudicial.

#A coluna Genero tem poucos faltantes (menos de 1%). Podemos optar por excluir essas linhas, pois a perda de dados é mínima e pode ajudar a manter a integridade do conjunto de dados para análise e modelagem, mas vamos focar na exclusão das linhas com Churn nulo, pois são inviáveis para o objetivo final. As outras colunas com valores ausentes podem ser tratadas posteriormente, dependendo da estratégia de imputação que decidirmos adotar.

# Removendo linhas onde a variável alvo 'Churn' está ausente
df = df.dropna(subset=['Churn'])

# Removendo linhas onde 'Genero' está ausente (perda mínima de dados)
df = df.dropna(subset=['Genero'])

print(f"Linhas restantes após a exclusão: {len(df)}")
```

C) Nessa etapa substitua pela média, mediana ou moda as linhas das colunas com valores faltantes que você analisou e entendeu que esse era o melhor método. Não se esqueça de justificar sua escolha.

**Dica variáveis numéricas:** Não se esqueça de verificar a distrbuição dos dados, média e mediana.

**Dica variáveis categóricas:** Caso precise substituir uma variável categórica utilize um gráfico de barras para verificar a que mais aparece e utilizar ela na substituição.
Ou você pode optar por excluir a coluna, mas justifique sua escolha.

Caso opte por substituir você pode utilizar a função fillna:
df['SUACOLUNA'].fillna("VALOR QUE SUBSTITUIRA O NULO", inplace=True)
Ela funcionará como a replace.

```python
#Pagamento_Mensal: É uma variável quantitativa contínua. Verificamos sua distribuição – normalmente possui assimetria (clientes com planos mais caros e mais baratos). A mediana é mais robusta a outliers do que a média. Portanto, usaremos a mediana para preencher os valores ausentes.

#Total_Pago: Também contínua, com forte assimetria (muitos clientes pagam pouco, alguns pagam muito). A mediana é novamente mais adequada.

#Genero: Variável categórica nominal. O método mais comum é substituir pela moda (valor que mais aparece), pois mantém a distribuição original.

# 1. Substituindo nulos em Pagamento_Mensal pela Mediana
mediana_pag = df['Pagamento_Mensal'].median()
df['Pagamento_Mensal'] = df['Pagamento_Mensal'].fillna(mediana_pag)

# 2. Substituindo nulos em Total_Pago pela Mediana
mediana_total = df['Total_Pago'].median()
df['Total_Pago'] = df['Total_Pago'].fillna(mediana_total)

# 3. Substituindo nulos em PhoneService pela Moda
moda_phone = df['PhoneService'].mode()[0]
df['PhoneService'] = df['PhoneService'].fillna(moda_phone)

# Verificação final: deve resultar em 0 para todas as colunas
print("Quantidade de nulos após o tratamento:")
print(df.isnull().sum())
```

# 3 - Verifique se encontramos valores digitados incorretamente, ou com letras maiusculas ou minusculas, ou até mesmo mesmos valores porém escritos de forma dirente e corrija.

```python
# Identificando colunas que contêm texto (object)
colunas_categoricas = df.select_dtypes(include=['object']).columns

# Aplicando str.lower() para padronizar e str.strip() para remover espaços extras
for col in colunas_categoricas:
    df[col] = df[col].str.lower().str.strip()

# Correção específica para a coluna Genero (mapeando variações como 'f' ou 'm')
mapa_genero = {'f': 'female', 'm': 'male', 'female': 'female', 'male': 'male'}
df['Genero'] = df['Genero'].map(mapa_genero)

# Verificando os valores únicos após a limpeza
print("Internet:", df['Servico_Internet'].unique())
print("Gênero:", df['Genero'].unique())
```

# [EXTRA] 4 - Caso deseje, para manter um padrão na sua base, renomeie as colunas ou deixando em português, ou inglês.

```python
# Renomeando as colunas para um padrão minúsculo e sem espaços
# Isso facilita a chamada das colunas (ex: df.pagamento_mensal em vez de df['Pagamento_Mensal'])
df.columns = [col.lower().replace(' ', '_') for col in df.columns]

# Exibindo os novos nomes das colunas
print("Novos nomes das colunas:")
print(df.columns)
```
